Εκτύπωση αποτελεσμάτων οργανωμένα

```python
import pandas as pd

experiments = {
    'YOLOv8n (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8n_soccernet/results.csv',
    'YOLOv8s (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_soccernet/results.csv',
    'YOLOv8m (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8m_soccernet/results.csv',
    'YOLOv8s (50%)':  '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_50pct-3/results.csv',
}

rows = []
for name, path in experiments.items():
    df = pd.read_csv(path)
    df.columns = df.columns.str.strip()
    best = df.loc[df['metrics/mAP50(B)'].idxmax()]
    rows.append({
        'Model': name,
        'Best Epoch': int(best['epoch']),
        'Precision': round(best['metrics/precision(B)'], 3),
        'Recall': round(best['metrics/recall(B)'], 3),
        'mAP50': round(best['metrics/mAP50(B)'], 3),
        'mAP50-95': round(best['metrics/mAP50-95(B)'], 3),
    })

results_df = pd.DataFrame(rows)
print(results_df.to_string(index=False))
```

Αξιολόγηση μοντέλων, TEST Set με συνολικές κατηγορίες. Εξαγωγή metrics, για κάθε κλάση.


```python


models = {
    'YOLOv8n (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8n_soccernet/weights/best.pt',
    'YOLOv8s (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_soccernet/weights/best.pt',
    'YOLOv8m (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8m_soccernet/weights/best.pt',
    'YOLOv8s (50%)':  '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_50pct-3/weights/best.pt',
}

rows = []
for name, path in models.items():
    model = YOLO(path)
    metrics = model.val(
        data='/content/drive/MyDrive/SoccerNet_YOLO/data.yaml',
        split='test',
        imgsz=640,
        batch=16,
        verbose=False
    )

    for i, cls in enumerate(['ball', 'person']):
        rows.append({
            'Model': name,
            'Class': cls,
            'Precision': round(float(metrics.box.p[i]), 3),
            'Recall': round(float(metrics.box.r[i]), 3),
            'mAP50': round(float(metrics.box.ap50[i]), 3),
            'mAP50-95': round(float(metrics.box.ap[i]), 3),
        })

df = pd.DataFrame(rows)
print(df.to_string(index=False))

# Save to CSV
df.to_csv('/content/drive/MyDrive/SoccerNet_YOLO/test_results_perclass.csv', index=False)
print("\nCSV Saved!")
```

```python


models = {
    'YOLOv8n (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8n_soccernet/weights/best.pt',
    'YOLOv8s (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_soccernet/weights/best.pt',
    'YOLOv8m (100%)': '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8m_soccernet/weights/best.pt',
    'YOLOv8s (50%)':  '/content/drive/MyDrive/SoccerNet_YOLO/runs/yolov8s_50pct-3/weights/best.pt',
}

for name, path in models.items():
    model = YOLO(path)
    metrics = model.val(
        data='/content/drive/MyDrive/SoccerNet_YOLO/data.yaml',
        split='test', imgsz=640, batch=16, verbose=False
    )
    print(f"{name}: P={metrics.box.mp:.3f}, R={metrics.box.mr:.3f}, mAP50={metrics.box.map50:.3f}, mAP50-95={metrics.box.map:.3f}")
```
